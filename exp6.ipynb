```python
import numpy as np
import pandas as pd
data = {
    'Country': ['India', 'USA', 'Japan', 'USA', 'Japan',
                'India', 'USA', 'India', 'Japan', 'India'],
    'Age': [25, 32, 28, 45, np.nan, 36, 29, 41, 50, 33],
    'Salary': [35000, 65000, 48000, 90000, 72000,
               55000, np.nan, 82000, 95000, 60000],
    'Purchased': ['Yes', 'No', 'Yes', 'No', 'Yes',
                  'No', 'Yes', 'No', 'Yes', 'No']
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

df['Country'] = df['Country'].fillna(df['Country'].mode()[0])

features = df.iloc[:, :-1].values
label = df.iloc[:, -1].values

from sklearn.impute import SimpleImputer

age = SimpleImputer(strategy='mean')
salary = SimpleImputer(strategy='mean')

features[:, [1]] = age.fit_transform(features[:, [1]])
features[:, [2]] = salary.fit_transform(features[:, [2]])

from sklearn.preprocessing import OneHotEncoder

oh = OneHotEncoder(sparse_output=False)
Country = oh.fit_transform(features[:, [0]])

print("\nEncoded Country:")
print(Country)

final_set = np.concatenate(
    (Country, features[:, [1, 2]]),
    axis=1
)

print("\nFinal Set:")
print(final_set)

from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
feat_standard_scaler = sc.fit_transform(final_set)

print("\nStandard Scaled Data:")
print(feat_standard_scaler)

from sklearn.preprocessing import MinMaxScaler

mms = MinMaxScaler(feature_range=(0, 1))
feat_minmax_scaler = mms.fit_transform(final_set)

print("\nMin-Max Scaled Data:")
print(feat_minmax_scaler)
```

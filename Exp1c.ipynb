```python
import pandas as pd
structured_data = pd.DataFrame({
'ID': [101, 102, 103],
'Name': ['Arthur', 'Thomas', 'Finn'],
'Age': [25, 30, 35]
})
print("Structured Data:\n", structured_data)
unstructured_data = "Python is a popular programming language used for data analysis, AF and web development."
print("\nUnstructured Data:\n", unstructured_data)
semi_structured_data = {'ID': 102, 'Name': 'Thomas', 'Attributes': {'Height': 175, 'Weight': 70}}
print("\nSemi-structured Data:\n", semi_structured_data)
```
